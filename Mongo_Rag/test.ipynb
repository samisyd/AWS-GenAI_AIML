```python
# import os
#https://www.mongodb.com/docs/atlas/atlas-vector-search/rag/?language-no-interface=python&embedding-model=voyage&llm=openai

import os
from dotenv import load_dotenv
load_dotenv()



```

```python
API_KEY = os.getenv("OPENAI_API_KEY")
MONGODB_URI = os.getenv("MONGODB_URI")
MONGODB_DB = os.getenv("MONGODB_DB", "sample_mflix")
MONGODB_COLLECTION = os.getenv("MONGODB_COLLECTION", "ragpdf")
MONGODB_VECTOR_INDEX = os.getenv("MONGODB_VECTOR_INDEX", "vector_index")
```

```python
from openai import OpenAI

# Initialize client
client = OpenAI(api_key=API_KEY)

# Specify the embedding model
model = "text-embedding-3-large"   # or "text-embedding-3-small"

# Define a function to generate embeddings
def get_embedding(text, input_type="document"):
    response = client.embeddings.create(
        model=model,
        input=text
    )
    print(response.data[0])
    return response.data[0].embedding
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Load the PDF
loader = PyPDFLoader("cold_and_flu_r.pdf")
data = loader.load()

# Split the data into chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=20)
documents = text_splitter.split_documents(data)
```

```python
# Prepare documents for insertion
docs_to_insert = [{
    "text": doc.page_content,
    "embedding": get_embedding(doc.page_content)
} for doc in documents]
```

```python
from pymongo import MongoClient

if not MONGODB_URI:
    raise ValueError("MONGODB_URI is not set. Add it to your .env file.")

# Connect to your MongoDB deployment
mongo_client = MongoClient(MONGODB_URI)
collection = mongo_client[MONGODB_DB][MONGODB_COLLECTION]

# Insert documents into the collection
result = collection.insert_many(docs_to_insert)
```

```python
result
```

```python
from pymongo.operations import SearchIndexModel
import time

# Create your index model, then create the search index
index_name = MONGODB_VECTOR_INDEX
search_index_model = SearchIndexModel(
  definition = {
    "fields": [
      {
        "type": "vector",
        "numDimensions": 3072,
        "path": "embedding",
        "similarity": "cosine"
      }
    ]
  },
  name = index_name,
  type = "vectorSearch"
)
collection.create_search_index(model=search_index_model)
```

```python
# Wait for initial sync to complete
print("Polling to check if the index is ready. This may take up to a minute.")
predicate=None
if predicate is None:
   predicate = lambda index: index.get("queryable") is True

while True:
   indices = list(collection.list_search_indexes(index_name))
   if len(indices) and predicate(indices[0]):
      break
   time.sleep(5)
print(index_name + " is ready for querying.")
```

```python
def get_query_results(query):

    # Convert query text → embedding vector
    query_embedding = get_embedding(query)

    pipeline = [
        {
            "$vectorSearch": {
                "index": "vector_index",
                "path": "embedding",
                "queryVector": query_embedding,
                "numCandidates": 150,   # typical range 100–500
                "limit": 5
            }
        },
        {
            "$project": {
                "_id": 0,
                "text": 1,
                "score": {"$meta": "vectorSearchScore"}
            }
        }
    ]

    results = collection.aggregate(pipeline)
    print(results)

    array_of_results = []
    for doc in results:
      array_of_results.append(doc)
    return array_of_results
```

```python
from openai import OpenAI

# Specify search query, retrieve relevant documents, and convert to string
query = "What are the Cold and flu symptoms"
context_docs = get_query_results(query)
context_string = " ".join([doc["text"] for doc in context_docs])

# Construct prompt for the LLM using the retrieved documents as the context
prompt = f"""Use the following pieces of context to answer the question at the end.
    {context_string}
    Question: {query}
"""

openai_client = OpenAI()

# OpenAI model to use
model_name = "gpt-4o"

completion = openai_client.chat.completions.create(
model=model_name,
messages=[{"role": "user",
    "content": prompt
  }]
)
print(completion.choices[0].message.content)
```
